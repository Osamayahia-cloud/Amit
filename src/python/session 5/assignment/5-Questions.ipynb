2. Write a program to print twin primes less than 1000. If two consecutive odd numbers are both prime then they are known as twin primes

```python
class Prog:
    
    def __init__(self,start,end):
        self.start = start
        self.end = end
        print("welcome to our program")
    def main_fun(self):
        for i in range(self.start, self.end, 2):
            #is prime
            prime1 = True
            for j in range(2, int(i ** 0.5) + 1):
                if i % j == 0:
                    prime1 = False
                    break

            #next odd num
            n = i + 2
            prime2 = True
            for j in range(2, int(n ** 0.5) + 1):
                if n % j == 0:
                    prime2 = False
                    break

            if prime1 and prime2:
                print(i, n)
                

x = Prog(3,1000)
x.main_fun()
```

3. Write a program to find out the prime factors of a number.
Example: prime factors of 56 - 2, 2, 2, 7

```python
class PrimeFa:

    def __init__(self, n):
        self.n = n

    def find_factors(self):
        i = 2

        while self.n > 1:
            if self.n % i == 0:
                print(i, end=" ")
                self.n = self.n // i
            else:
                i += 1


n = int(input("Enter a number: "))

x = PrimeFa(n)
x.find_factors()
```

4. Write a function that converts a decimal number to binary
number

```python
class Binary:

    def __init__(self, n):
        self.n = n

    def to_binary(self):
        binary = ""

        while self.n > 0:
            remainder = self.n % 2
            binary = str(remainder) + binary
            self.n = self.n // 2

        return binary


num = int(input("Enter a decimal number: "))

x = Binary(num)

print(x.to_binary())
```

5. A number is called perfect if the sum of proper divisors of that
number is equal to the number. For example 28 is perfect
number, since 1+2+4+7+14=28. Write a program to print all the
perfect numbers in a given range

```python
class PerfectNumber:

    def __init__(self, start, end):
        self.start = start
        self.end = end

    def perfect_numbers(self):

        for num in range(self.start, self.end + 1):
            total = 0

            for i in range(1, num):
                if num % i == 0:
                    total += i

            if total == num:
                print(num)


start = int(input("Enter start: "))
end = int(input("Enter end: "))

x = PerfectNumber(start, end)
x.perfect_numbers()
```
