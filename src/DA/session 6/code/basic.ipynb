```python
class person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def __str__(self):
        return f"my name is {self.name},age is {self.age}"
    def __repr__(self):
        return f"my name is {self.name},age is {self.age}"
```

```python
s1 = person("osaam",20)
str(s1)
```
