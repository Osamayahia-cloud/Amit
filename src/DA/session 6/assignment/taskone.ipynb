```python
import numpy as np


def fact(mode, col, row, value=None):

    if mode == "zeros":
        return np.zeros((col, row))

    elif mode == "ones":
        return np.ones((col, row))

    elif mode == "full":
        if value is None:
            raise ValueError("Value must be provided for 'full' mode.")
        return np.full((col, row), value)

    elif mode == "identity":
        return np.eye(col)

    else:
        print("Invalid mode. Choose: zeros, ones, full, identity.")


print(fact("zeros", 2, 3))
```
